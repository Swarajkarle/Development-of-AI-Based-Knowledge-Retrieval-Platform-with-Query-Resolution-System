### Q1. What is Python?

**Answer:**

Python is a high-level, interpreted, general-purpose programming language.
It is widely used for web development, data analysis, artificial intelligence,
machine learning, automation, scripting, and software development.

**Key Features:**
- Easy to learn and read
- Interpreted language
- Dynamically typed
- Object-oriented
- Large standard library
- Supports multiple programming paradigms
- Large community and ecosystem

**Interview Tip:**

If the interviewer asks "Why Python?", mention its simple syntax,
large library ecosystem, rapid development, and applications in
Data Science, AI/ML, automation, and backend development.

### Q3. What is the difference between compiled and interpreted languages?

**Answer:**

A compiled language is generally translated into machine code before
execution, while an interpreted language is generally executed through
an interpreter.

**Python:**

Python is commonly described as an interpreted language. Python source
code is first compiled into bytecode, which is then executed by the
Python Virtual Machine (PVM).

**Interview Tip:**

A good answer is:

"Python is interpreted at the language level, although CPython first
compiles the source code into bytecode (.pyc), which is executed by
the Python Virtual Machine."

### Q4. What is a variable in Python?

**Answer:**

A variable is a name that refers to an object in memory.

Example:

```python
name = "Swaraj"
age = 20
cgpa = 8.5

print(name)
print(age)
print(cgpa)
```

**Output:**

Swaraj
20
8.5

**Interview Tip:**

Python variables do not require explicit declaration of a data type.
The type is associated with the object referenced by the variable.
